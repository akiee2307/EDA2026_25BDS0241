```python
# 1. Variables and print
name = "Alex"
age = 25
print(f"{name} is {age} years old")
```

```python
# 2. If-else conditional
num = 10
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# 3. For loop
for i in range(5):
    print(i)
```

```python
# 4. While loop
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
# 5. List and list operations
fruits = ["apple", "banana", "cherry"]
fruits.append("orange")
print(fruits)
```

```python
# 6. Dictionary
person = {"name": "Alex", "age": 25}
print(person["name"])
```

```python
# 7. Function definition
def add(a, b):
    return a + b
print(add(3, 4))
```

```python
# 8. String manipulation
text = "hello world"
print(text.upper())
print(text.split())
```

```python
# 9. List comprehension
squares = [x**2 for x in range(10)]
print(squares)
```

```python
# 10. Try-except error handling
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```
